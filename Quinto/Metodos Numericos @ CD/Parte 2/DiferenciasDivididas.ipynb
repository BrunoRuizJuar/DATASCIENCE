```julia
using Printf

"""
    diferencias_divididas(x::Vector{Float64}, y::Vector{Float64})

Calcula los coeficientes del polinomio de interpolación de Newton
usando el método de diferencias divididas, como se describe en la imagen.

# Argumentos
- `x::Vector{Float64}`: Vector con las coordenadas x de los puntos (nodos).
- `y::Vector{Float64}`: Vector con las coordenadas y de los puntos (valores de la función).

# Retorno
- `Vector{Float64}`: Un vector que contiene los coeficientes calculados
  (F₀,₀, F₁,₁, ..., Fₙ,ₙ).
"""
function diferencias_divididas(x::Vector{Float64}, y::Vector{Float64})
    n = length(x)
    if n != length(y)
        error("Los vectores 'x' e 'y' deben tener la misma longitud.")
    end

    # Se crea una copia de 'y' para no modificar el vector original.
    # Este vector 'c' almacenará los coeficientes.
    c = copy(y)

    # El bucle externo 'j' itera sobre las columnas de la tabla de diferencias
    # (es decir, el grado de la diferencia).
    for j in 2:n
        # El bucle interno 'i' calcula cada diferencia dividida.
        # Se itera hacia atrás para usar los valores de la iteración anterior
        # antes de que sean sobreescritos.
        for i in n:-1:j
            c[i] = (c[i] - c[i-1]) / (x[i] - x[i-j+1])
        end
    end

    return c
end
```

```julia
"""
    evaluar_polinomio(coefs::Vector{Float64}, x::Vector{Float64}, t::Float64)

Evalúa el polinomio de Newton en un punto `t` usando la forma anidada
(método de Horner) para mayor eficiencia y estabilidad numérica.

El polinomio es de la forma:
P(t) = c₀ + c₁(t-x₀) + c₂(t-x₀)(t-x₁) + ...

# Argumentos
- `coefs::Vector{Float64}`: Coeficientes del polinomio (obtenidos de `diferencias_divididas`).
- `x::Vector{Float64}`: Vector original de nodos.
- `t::Float64`: El punto en el cual se desea evaluar el polinomio.

# Retorno
- `Float64`: El valor del polinomio P(t).
"""

function evaluar_polinomio(coefs::Vector{Float64}, x::Vector{Float64}, t::Float64)
    n = length(coefs)
    # Empezamos con el último coeficiente
    resultado = coefs[n]

    # Aplicamos la forma anidada (método de Horner)
    for i in (n-1):-1:1
        resultado = resultado * (t - x[i]) + coefs[i]
    end

    return resultado
end
```

```julia
# --- Ejemplo de Uso ---

# Datos de entrada (puntos a interpolar)
x_puntos = [1.0, 1.3, 1.6, 1.9, 2.2]
y_puntos = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]

# 1. Obtener los coeficientes según el algoritmo
coeficientes = diferencias_divididas(x_puntos, y_puntos)

println("Puntos de entrada:")
println("x = ", x_puntos)
println("y = ", y_puntos)
println("-"^30)

println("Coeficientes del Polinomio de Newton (F₀,₀, F₁,₁, ...):")
for (i, coef) in enumerate(coeficientes)
    @printf "F_%d,%d = %+.8f\n" (i-1) (i-1) coef
end
println("-"^30)


# 2. Evaluar el polinomio en un punto, por ejemplo t = 1.5
valor_interpolado = evaluar_polinomio(coeficientes, x_puntos, 1.5)

println("Evaluación del polinomio en t = 1.5:")
@printf "P(1.5) = %.8f\n" valor_interpolado
```
