```julia
"""
    ajuste_trigonometrico_discreto(f, n, m)

Calcula los coeficientes para la Aproximación Trigonométrica de Mínimos Cuadrados
usando el Teorema 8.13 del texto.

Argumentos:
- `f`: Función a aproximar (debe aceptar valores en [-π, π]).
- `n`: Grado del polinomio trigonométrico.
- `m`: Define la cantidad de nodos 2m (debe cumplirse que 2m > n).

Retorna:
- `a`: Vector de coeficientes coseno [a_0, ..., a_n]
- `b`: Vector de coeficientes seno [b_1, ..., b_{n-1}]
"""
function ajuste_trigonometrico_discreto(f::Function, n::Int, m::Int)
    if n >= m
        error("El grado n debe ser menor que m.")
    end

    #Generación de Nodos x_j y valores y_j
    N_puntos = 2 * m
    x = zeros(N_puntos)
    y = zeros(N_puntos)
    
    for j in 0:(N_puntos - 1)
        x[j+1] = -π + (j / m) * π
        y[j+1] = f(x[j+1])
    end

    #Coeficientes a_k 
    a = zeros(n + 1) 
    for k in 0:n
        suma = 0.0
        for j in 0:(N_puntos - 1)
            suma += y[j+1] * cos(k * x[j+1])
        end
        a[k+1] = (1 / m) * suma
    end

    #Coeficientes b_k
    b = zeros(n - 1)
    if n > 1
        for k in 1:(n - 1)
            suma = 0.0
            for j in 0:(N_puntos - 1)
                suma += y[j+1] * sin(k * x[j+1])
            end
            b[k] = (1 / m) * suma
        end
    end

    return a, b
end
```

```julia
"""
    evaluar_trigonometrico(x_eval, a, b, n)

Evalúa el polinomio S_n(x) en el punto x_eval usando los coeficientes calculados.
Fórmula: S_n(x) = a_0/2 + a_n*cos(nx) + sum(a_k*cos(kx) + b_k*sin(kx))
"""
function evaluar_trigonometrico(x_eval, a::Vector, b::Vector, n::Int)
    # Término inicial a_0 / 2
    valor = a[1] / 2.0
    
    # Término final a_n * cos(nx)
    valor += a[n+1] * cos(n * x_eval)
    
    # Suma k = 1 hasta n-1
    for k in 1:(n - 1)
        term_cos = a[k+1] * cos(k * x_eval)
        term_sin = b[k] * sin(k * x_eval)
        valor += term_cos + term_sin
    end
    
    return valor
end
```

```julia
#funcion a aproximar
f_ejemplo(x) = 2*x^2 - 9

n = 2
m = 3

#Calcular coeficientes
a_calc, b_calc = ajuste_trigonometrico_discreto(f_ejemplo, n, m)

println("Coeficientes:")
println("a_0: $(a_calc[1])")
println("a_1: $(a_calc[2])")
println("a_2: $(a_calc[3])")

if length(b_calc) > 0
    println("b_1: $(b_calc[1])")
else
    println("b_1: No aplica")
end

# 4. Evaluar en un punto de prueba (ej. x = 0)
# f(0) = -9
val_aprox = evaluar_trigonometrico(0.0, a_calc, b_calc, n)
println("\nx=0:")
println("Real:      $(f_ejemplo(0.0))")
println("Aproximado: $val_aprox")
```
